```python
text1=input()
text2=input()

text1=text1.lower()
text2=text2.lower()

if(len(text1)<len(text2) and text1 in text2):
  print(text1,end="")
elif(len(text2)<len(text1) and text2 in text1):
  print(text2,end="")
else:
  print("None")
```
