```python
#Spiral Matrix
# --------Input--------
# 123
# 456
# 789

#-------Output---------
#123698745

row,col=map(int,input().split())
matrix=[]
for i in range(row):
  matrix.append(list(map(int,input().split())))

top=0
bottom=row-1
left=0
right=col-1
while(top<=bottom and left<=right):
  for i in range(left, right + 1):
        print(matrix[top][i], end=" ")
  top += 1
  for i in range(top,bottom+1):
    print(matrix[i][right],end=" ")
  right-=1


  if top <= bottom:
    for i in range(right, left - 1, -1):
      print(matrix[bottom][i], end=" ")
    bottom -= 1


    if left <= right:
        for i in range(bottom, top - 1, -1):
            print(matrix[i][left], end=" ")
        left += 1
```

```python
arr=list(map(int,input().split()))
def count_bits(n):
  count=0
  while n>0:
    count+=n%2
    n=n//2
  return count

arr.sort(key=count_bits)
print(*arr)
```

```python
#K numbers Minimize the difference
N=int(input())
arr=list(map(int,input().split()))
K=int(input())
arr.sort()
minimum=float('inf')

for i in range(N - K + 1):
    difference = arr[i + K - 1] - arr[i]

    if difference < minimum:
        minimum = difference

print(minimum)
```

```python
arr=list(map(int,input().split()))
result=0
for i in range(len(arr) - 1):
  result^=i
diff_bit = result & (-result)


a = 0
b = 0
for num in arr:
    if num & diff_bit:
        a ^= num
    else:
        b ^= num

result1 = sorted([a, b])
print(*result1)
```

```python
#Identical Block
n=int(input())
arr=[]
for i in range(n):
  element=int(input())
  arr.append(element)

j=0
count=0
while(j<n):
  current_block=arr[j]
  count_block=0
  while(j<n and arr[j]==current_block):
    count_block+=1
    j+=1
  if(current_block==count_block):
    count+=1
print(count)
```

```python
#return the sum of prime numbers in given range
def Primerange(n,m):
  sum=0
  for i in range(n,m+1):
    count=0
    for j in range(1,i+1):
      if(i%j==0):
        count+=1
    if(count==2):
      sum+=i
  return sum

n=int(input())
m=int(input())
print(Primerange(n,m))
```
