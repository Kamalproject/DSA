```python
#Spiral Matrix
# --------Input--------
# 123
# 456
# 789

#-------Output---------
#123698745

row,col=map(int,input().split())
matrix=[]
for i in range(row):
  matrix.append(list(map(int,input().split())))

top=0
bottom=row-1
left=0
right=col-1
while(top<=bottom and left<=right):
  for i in range(left, right + 1):
        print(matrix[top][i], end=" ")
  top += 1
  for i in range(top,bottom+1):
    print(matrix[i][right],end=" ")
  right-=1


  if top <= bottom:
    for i in range(right, left - 1, -1):
      print(matrix[bottom][i], end=" ")
    bottom -= 1


    if left <= right:
        for i in range(bottom, top - 1, -1):
            print(matrix[i][left], end=" ")
        left += 1
```

```python
def findDifference(m,n):
  four_Sum=0
  seven_Sum=0
  while(m<=n):
    if(m%4==0):
      digit=m//10
      reminder=m%10
      four_Sum+=(digit+reminder)
    elif(m%7==0):
      digit=m//10
      reminder=m%10
      seven_Sum+=(digit+reminder)
    m+=1
    diff=four_Sum-seven_Sum
  return diff

m=int(input())
n=int(input())
findDifference(m,n)
```

```python
#Find the Second Largest element in given array
def Second_largest(arr):
  second_largest=arr[0]
  largest=arr[0]
  for i in range(len(arr)):
    if(arr[i]>largest):
      second_largest=largest
      largest=arr[i]
    elif(arr[i]>second_largest and arr[i]!=largest):
      second_largest=arr[i]
  return second_largest

arr=[12,35,1,10,34,1]
Second_largest(arr)
```

```python
#Divisible sum difference

def DivisibleSum(N,M):
  divisible_sum=0
  notdivisible_sum=0
  diff=0
  for i in range(1,N+1):
    if(i%M==0):
      divisible_sum+=i
    else:
      notdivisible_sum+=i
  if(divisible_sum>notdivisible_sum):
    diff= divisible_sum-notdivisible_sum
  else:
    diff= notdivisible_sum-divisible_sum
  return diff

N,M=map(int,input().split())
DivisibleSum(N,M)
```

```python
#Aramstrong number
def armstrongNumber(number):
  text=str(number)
  temp=number
  sum=0
  while(temp>0):
    digit=temp%10
    sum+=digit**len(text)
    temp//=10
  if(sum==number):
    print("Armstrong Number")
  else:
    print("Not an Armstrong Number")
N=int(input())
armstrongNumber(N)
```

```python
#GCD using Uclidian Algorithm

def GCD(a,b):

  while(b!=0):
    c=a%b
    a=b
    b=c
  return a
x,y=map(int,input().split())
GCD(x,y)
```

```python
#Palindrome String
def Palindrome(name):
  name=list(name)
  flag=True
  for i in range(len(name)):
    for j in range(len(name)-1,-1,-1):
      if(name[i]!=name[j]):
        flag=False
        break

      i+=1
      j-=1
    break
  if(flag):
    print("Palindrome")
  else:
    print("Not a palindrome")

text=input()
Palindrome(text)
```

```python
def TeamGame(N):
  arr=[]
  for i in range(N):
    element=input().upper()
    arr.append(element)
  count_a=0
  count_b=0
  for i in range(len(arr)):
    if(arr[i]=='A'):
      count_a+=1
    else:
      count_b+=1
  if(count_a>count_b):
    print("Team A")
  else:
    print("Team B")

TeamGame(5)
```

```python
#Move Zeros to end
def MoveZeros(arr):
  left=0
  right=0
  while(left<len(arr)):
    if(arr[left]!=0):
      temp=arr[right]
      arr[right]=arr[left]
      arr[left]=temp
      left+=1
      right+=1
    else:
      left+=1
  return tuple(arr)
arr=[0,1,0,3,12]
MoveZeros(arr)
```
