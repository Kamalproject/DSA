```python
#Sufficient Food for Rats
print("Rat number: ")
N=int(input())
print("UNIT")
unit=int(input())
arr=list(map(int,input().split()))
sum=0
required=N*unit
for i in range(len(arr)):
  sum+=arr[i]
  if(required<=sum):
    print(i+1)
    break
else:
  print(-1)
```

```python
#Average of numbers divisible by 3
arr=list(map(int,input().split()))
count=0
sum=0
for i in arr:
  if(i%3==0):
    count+=1
    sum+=i
if(sum%count==0):
  print(sum//count)
else:
  print(sum/count)
```

```python
#Spiral Matrix
# --------Input--------
# 123
# 456
# 789

#-------Output---------
#123698745

row,col=map(int,input().split())
matrix=[]
for i in range(row):
  matrix.append(list(map(int,input().split())))

top=0
bottom=row-1
left=0
right=col-1
while(top<=bottom and left<=right):
  for i in range(left, right + 1):
        print(matrix[top][i], end=" ")
  top += 1
  for i in range(top,bottom+1):
    print(matrix[i][right],end=" ")
  right-=1


  if top <= bottom:
    for i in range(right, left - 1, -1):
      print(matrix[bottom][i], end=" ")
    bottom -= 1


    if left <= right:
        for i in range(bottom, top - 1, -1):
            print(matrix[i][left], end=" ")
        left += 1
```

```python
#Count even/odd index-value combinations
#It will check the both index and value at that index
#if meet the condition then it update the count
arr=list(map(int,input().split()))
count=0
for i in range(len(arr)):
  if(i%2==0 and arr[i]%2==0 or i%2!=0 and arr[i]%2!=0):
    count+=1
print(count)
```

```python
#Count numbers Ending in 2,5 or 7
#Basically it check the unitdigit
#if unitdigit contains 2,5,7 then return the count

arr=list(map(int,input().split()))
count=0
for i in range(len(arr)):
  digit=arr[i]%10
  if(digit==2 or digit==5 or digit==7):
    count+=1

print(count)
```

```python
arr=list(map(int,input().split()))
def count_bits(n):
  count=0
  while n>0:
    count+=n%2
    n=n//2
  return count

arr.sort(key=count_bits)
print(*arr)
```

```python
#K numbers Minimize the difference
N=int(input())
arr=list(map(int,input().split()))
K=int(input())
arr.sort()
minimum=float('inf')

for i in range(N - K + 1):
    difference = arr[i + K - 1] - arr[i]

    if difference < minimum:
        minimum = difference

print(minimum)
```
