```python
text1=input()
text2=input()

text1=text1.lower()
text2=text2.lower()

if(len(text1)<len(text2) and text1 in text2):
  print(text1,end="")
elif(len(text2)<len(text1) and text2 in text1):
  print(text2,end="")
else:
  print("None")
```

```python
#Sufficient Food for Rats
print("Rat number: ")
N=int(input())
print("UNIT")
unit=int(input())
arr=list(map(int,input().split()))
sum=0
required=N*unit
for i in range(len(arr)):
  sum+=arr[i]
  if(required<=sum):
    print(i+1)
    break
else:
  print(-1)
```

```python
#Average of numbers divisible by 3
arr=list(map(int,input().split()))
count=0
sum=0
for i in arr:
  if(i%3==0):
    count+=1
    sum+=i
print(sum//count)
```
